```python
import json
import pprint
import re
```

```python
with open(r"C:\Users\mdabe\Downloads\TableExtraction_4de36193-616d-4885-9fba-0c41611faaa4.csv", "r") as f:
    data = f.read()
```

```python
text = ""
for line in data.split("\n"):
    if line.startswith("\"") and line.endswith("\""):
        line = line[1:-1]
        
    text += line + "\n"
```

```python
lines = text.split("\n")
functions = {}
category = ''
for i, line in enumerate(lines):
    if i + 1 < len(lines) and lines[i + 1] == "Name Description": # categoria
        category = line
        continue

    if line == "Name Description": # pulando a linha de cabeçalho
        continue

    # se a linha for toda maiúscula e uma palavra só, é função
    # se a primeira palavra for maiúscula seguida de um "(" é função
    if line.isupper() or re.match(r"^[A-Z]+\(", line):
        if line.isupper():
            name = line
            definition = line
            varnames = []
        else:
            name, vars_ = line.split("(", 1)
            definition = line
            varnames = vars_[:-1].split(",")
            varnames = [var.strip() for var in varnames]

        description = lines[i + 1]

        functions[name] = {
            "category": category,
            "name": name,
            "definition": definition,
            "varnames": varnames,
            "description": description,
            "close" : ""
        }
```

```python
with open("functions.json", "w") as f:
    json.dump(functions, f, indent=4)
```
